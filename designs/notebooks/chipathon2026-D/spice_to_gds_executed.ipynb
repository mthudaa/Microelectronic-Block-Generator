# SPICE → GDS Layout Generator
SSCS Chipathon 2026 — gLayout Track

Converts SPICE subcircuit netlist to DRC-clean GDSII layout.
Includes SVG preview, AC simulation, DRC/LVS/PEX verification.

**PDK**: gf180mcuD  |  **Supply**: 1.8 V

```python
# ==========================================
# 1. Define SPICE Netlist
# ==========================================
# Edit this cell to change your circuit.
# Supports: nfet_03v3, pfet_03v3 (MOSFETs)
#          ppolyf_u, npolyf_u, rm1-4  (Resistors)
#          cap_mim_2f0fF, cap_nmos_03v3 (Capacitors)

netlist = """
.lib "/home/huda/.volare/gf180mcuD/libs.tech/ngspice/sm141064.ngspice" typical
.subckt opa_tuned3 vdd vss iref vip vin vout
XM1 net1 net1 vdd vdd pfet_03v3 L=0.5u W=2u nf=1 m=1
XM2 net2 net1 vdd vdd pfet_03v3 L=0.5u W=2u nf=1 m=1
XM3 net1 vip net3 net3 nfet_03v3 L=0.5u W=2u nf=1 m=1
XM4 net2 vin net3 net3 nfet_03v3 L=0.5u W=2u nf=1 m=1
XM5 net3 iref vss vss nfet_03v3 L=0.5u W=2u nf=1 m=1
XM6 vout net2 vdd vdd pfet_03v3 L=0.5u W=3u nf=1 m=1
XM7 vout iref vss vss nfet_03v3 L=0.5u W=2u nf=1 m=1
XM8 iref iref vss vss nfet_03v3 L=0.5u W=2u nf=1 m=1
.ends
"""
```

```python
# ==========================================
# 2. Generate Layout + SVG Preview
# ==========================================
import gdstk
from IPython.display import SVG, display
from core import spice_to_gds

result = spice_to_gds(netlist, mode='analog', add_labels=True)
result.write_gds('out.gds')

# Generate SVG
lib = gdstk.read_gds('out.gds')
lib.top_level()[0].write_svg('out.svg')
display(SVG('out.svg'))
print(f'Layout: out.gds  |  SVG: out.svg')
bb = lib.top_level()[0].bounding_box()
print(f'Size: {bb[1][0]-bb[0][0]:.0f} x {bb[1][1]-bb[0][1]:.0f} um')
```

```python
# ==========================================
# 3. Verify GDS
# ==========================================
from core.checks import validate_gds
v = validate_gds('out.gds')
print(v['message'])
```

```python
# ==========================================
# 4. DRC (requires Magic)
# ==========================================
from core.checks import run_drc, check_tools
tools = check_tools()
if tools['magic']:
    drc = run_drc('out.gds', cell_name='opa_tuned3')
    print(drc['summary'])
else:
    print('Skipping DRC — Magic not available')
```

```python
# ==========================================
# 5. With Resistor + Capacitor + SVG
# ==========================================
netlist_rc = '''
.lib /home/huda/.volare/gf180mcuD/libs.tech/ngspice/sm141064.ngspice typical
.subckt rc_filter vdd vin vout vss
M1 vout vin vss vss nfet_03v3 W=1u L=0.28u nf=1
M2 vdd vin vout vdd pfet_03v3 W=1u L=0.28u nf=1
R1 vout vdd ppolyf_u W=5u L=10u
C1 vout vss cap_mim_2f0fF W=10u L=10u
.ends
'''
result2 = spice_to_gds(netlist_rc, mode='analog', add_labels=True)
result2.write_gds('rc_filter.gds')

lib2 = gdstk.read_gds('rc_filter.gds')
lib2.top_level()[0].write_svg('rc_filter.svg')
display(SVG('rc_filter.svg'))
print(f'Layout: rc_filter.gds  |  SVG: rc_filter.svg')
bb2 = lib2.top_level()[0].bounding_box()
print(f'Size: {bb2[1][0]-bb2[0][0]:.0f} x {bb2[1][1]-bb2[0][1]:.0f} um')
```

```python
# ==========================================
# 6. LVS (requires Magic + netgen)
# ==========================================
from core.checks import run_lvs
lvs = run_lvs('out.gds', netlist_content=netlist, cell_name='opa_tuned3',
              workdir='/tmp/lvs_notebook', auto_fix_ports=True)
print(f"LVS: {'MATCHED' if lvs['match'] else 'MISMATCHED'}")
if not lvs['match']:
    print(f"  {lvs['summary']['message']}")
```
